```python
import pandas as pd
df = pd.read_csv("../data/coches.csv", sep=";")
df.shape, df.columns.tolist()
```

```python
df.head(10)
df.dtypes
df.isna().sum()          # ¿cuántos vacíos hay por columna?
```

```python
df["brand"].value_counts().head(15)      # marcas más frecuentes
df["fuel"].value_counts()
```

```python
# Los precios y km suelen venir como texto ("12.500 €"): conviértelos para analizarlos
precio = df["price (eur)"].astype(str).str.replace(r"[^\d]", "", regex=True).astype(float)
precio.describe()
precio.plot.hist(bins=40, title="Distribución de precios")
```

```python
anio = pd.to_numeric(df["year"], errors="coerce")
pd.DataFrame({"anio": anio, "precio": precio}).plot.scatter(x="anio", y="precio", alpha=.4)
```
